```python
#1.b The minimal fix
def correct_add_function(arg1, arg2):
    arg1_index = 0
    while arg1_index < len(arg1):
        arg_2_sum = 0
        for arg2_elements in arg2:
            arg_2_sum += arg2_elements                     # changed: accumlate
        arg1[arg1_index] = arg1[arg1_index] + arg_2_sum    # changed: add to arg1
        arg1_index += 1
    return arg1

correct_add_function([1, 2, 3], [1, 1, 1])
```

```python
def exception_add_function(arg1, arg2):
    try:
        expected_type = type(arg1[0])       # the type every element should share
        for arg_number, arg in enumerate([arg1, arg2], start=1):
            for n, element in enumerate(arg):
                if type(element) != expected_type:
                    raise TypeError(
                        f"Your input argument {arg_number} at element {n} "
                        f"is not of the expected type. Please change this and rerun.")
    except TypeError as error:
        return str(error)

    #numeric section
    if sum([type(i)==int for i in arg1])==len(arg1) and \
       sum([type(i)==int for i in arg2])==len(arg2):
        arg1_index=0
        while arg1_index < len(arg1):
            ar_2_sum = 0
            for arg2_elements in arg2:
                arg_2_sum += arg2_elements
            arg1[arg1_index]=arg1[arg1_index]+arg_2_sum
            arg1_index+=1
        return arg1
    #string section
    elif sum([type(i)==str for i in arg1])==len(arg1) and \
       sum([type(i)==str for i in arg2])==len(arg2):
        arg_index=0
        while arg1_index < len(arg1):
            arg_2_sum = ''
            for arg2_elements in arg2:
                arg_2_sum += arg2_elements
            arg1[arg1_index]=arg1[arg1_index]+str(arg_2_sum)
            arg1_index+=1
        return arg1

arg_str_1 = ['1', '2', '3']
arg_str_2 = ['1', '1', 1]
print(exception_add_function(arg_str_1, arg_str_2))
print(exception_add_function(["5", "2", 5], ["1", "1"]))
```

```python
#2.c Correcting invalid input
def correction_add_function(arg1, arg2):
    try:
        # mixed types (e.g. ints and strings together) can't go down either branch
        if len(set(type(i) for i in arg1 + arg2)) > 1:
            raise TypeError
    except TypeError:
        # convert copies inside the function; the caller's lists are untouched
        arg1 = [str(i) for i in arg1]
        arg2 = [str(i) for i in arg2]

    #numeric section
    if sum([type(i)==int for i in arg1])==len(arg1) and \
       sum([type(i)==int for i in arg2])==len(arg2):
        arg1_index=0
        while arg1_index < len(arg1):
            ar_2_sum = 0
            for arg2_elements in arg2:
                arg_2_sum += arg2_elements
            arg1[arg1_index]=arg1[arg1_index]+arg_2_sum
            arg1_index+=1
        return arg1
    #string section
    elif sum([type(i)==str for i in arg1])==len(arg1) and \
       sum([type(i)==str for i in arg2])==len(arg2):
        arg1_index=0
        while arg1_index < len(arg1):
            arg_2_sum = ''
            for arg2_elements in arg2:
                arg_2_sum += arg2_elements
            arg1[arg1_index]=arg1[arg1_index]+str(arg_2_sum)
            arg1_index+=1
        return arg1

arg_str_1 = ['1', '2', '3']
arg_str_2 = ['1', '1', 1]
print(correction_add_function(arg_str_1, arg_str_2))
print("Original inputs unchanged:", arg_str_1, arg_str_2)
```
